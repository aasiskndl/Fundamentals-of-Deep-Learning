```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="bigscience/bloom-560m",
    max_new_tokens=150,
    temperature=0.7,
    do_sample=True
)
```

```python
prompt = "what is supervised learning in machine learning and AI"
output = pipe(prompt)[0]['generated_text']
print(output)
```
